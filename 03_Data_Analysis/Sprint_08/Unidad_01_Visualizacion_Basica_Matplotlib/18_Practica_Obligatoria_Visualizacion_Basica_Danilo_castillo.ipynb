#### #1.2  

Recrea la siguiente figura, para ello construye en apartados separados cada gráfico y luego haz un apartado en el que se junten los cuatro.  

Las figuras son:
1. Las ciudades más pobladas, con los numeros de las poblaciones (tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center"). (Colores: azul, gris, aceituna, marrón, con trasparencia)  

2. Histograma del área total ocupada por las ciudades. (50 bins, fuente = "Bradley Hand ITC", tamaño 18, color azul acero o "steelblue")
3. El scatter de área de agua de un ciudad con su poblacion.
4. El boxplot de la altura de las ciudades.

```python
# ciudades más pobladas

top = df.sort_values("population_total", ascending=False).head(4)

# Valores
ciudades = top["city"].values
poblaciones = top["population_total"].values

# Colores solicitados
colores = ["blue", "gray", "olive", "brown"]

plt.figure(figsize=(12, 6))

# Gráfico de barras
bars = plt.bar(ciudades, poblaciones, color=colores, alpha=0.6)

# Añadir texto centrado sobre cada barra
for i, (ciudad, poblacion) in enumerate(zip(ciudades, poblaciones)):
    plt.text(
        i, poblacion + poblacion*0.02,  # pequeño desplazamiento hacia arriba
        f"{poblacion:,}",               # formato con separador de miles
        ha="center", va="bottom",
        fontsize=12)

plt.title("Ciudades más pobladas de California")
plt.ylabel("Habitantes miles")

plt.tight_layout()
plt.show();
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df["area_total_km2"], bins=50, alpha=0.6, label="Hist Area")  # transparencia

# títulos y ejes con formato
plt.xlabel("Superficie ($km^2$)", fontsize=18)
plt.ylabel("Número de Ciudades", fontsize=18)

# formato

font_dict = {
    "family" : "Bradley Hand ITC",
    "color" : "steelblue"}
# leyendas
plt.legend()

plt.text(fontdict = font_dict)

plt.show()
```

```python
# Scatter plot
plt.figure(figsize=(10, 8))

plt.scatter(
    df["area_water_km2"],
    df["population_total"] / 1000,  
    color="green",
    alpha=0.6)

# titulos 
plt.title("Población por kilometro cuadrado de agua", fontsize=14)
plt.xlabel("Área de agua en la ciudad ($km^2$)", fontsize=12)
plt.ylabel("Población (miles habitantes)", fontsize=12)

# outliers
outliers = df[
    (df["area_water_km2"] > 80) & 
    (df["population_total"] > 3000000)
]

# anotaciones
for _, row in outliers.iterrows():
    plt.annotate(
        "Outliers",
        xy=(row["area_water_km2"], row["population_total"] / 1000),
        xytext=(row["area_water_km2"] - 30, row["population_total"] / 1000 + 500),
        arrowprops=dict(arrowstyle="->", lw=1),
        fontsize=10
    )

plt.show()
```

```python

plt.figure(figsize=(12, 6))

plt.boxplot(
    df["elevation_m"],
    vert=False,
    patch_artist=True,                 
    boxprops=dict(facecolor="red"), 
    medianprops=dict(color="black"),   
    whiskerprops=dict(color="black"),
    capprops=dict(color="black"),
    flierprops=dict(marker='o', markersize=6, markerfacecolor="white")
)

# Títulos y etiquetas
plt.title("Diagrama de caja de la Altitud de las ciudades de California", fontsize=14)
plt.xlabel("Altitud (m)", fontsize=12)
plt.show()
```

```python
# TODOS JUNTOS

#plt.figure()
#fig,ax = plt.subplots(2,2, sharex = "col", sharey = "row")

top = df.sort_values("population_total", ascending=False).head(4)
ciudades = top["city"].values
poblaciones = top["population_total"].values
colores = ["blue", "gray", "olive", "brown"]

font_dict = {
    "family": "Bradley Hand ITC",
    "color": "steelblue"
}

# 
fig, axes = plt.subplots(2, 2, figsize=(18, 12))
(ax1, ax2), (ax3, ax4) = axes


# ciudades mas pobladas
bars = ax1.bar(ciudades, poblaciones, color=colores, alpha=0.6)

for i, (ciudad, poblacion) in enumerate(zip(ciudades, poblaciones)):
    ax1.text(
        i, poblacion + poblacion*0.02,
        f"{poblacion:,}",
        ha="center", va="bottom", fontsize=10
    )

ax1.set_title("Ciudades más pobladas de California")
ax1.set_ylabel("Habitantes (miles)")

#superficie en histograma

ax2.hist(df["area_total_km2"], bins=50, alpha=0.6, label="Hist Area")
ax2.set_xlabel("Superficie ($km^2$)", fontsize=12)
ax2.set_ylabel("Número de Ciudades", fontsize=12)
ax2.legend()
ax2.text(0.95, 0.95, "", fontdict=font_dict, ha="right", va="top", transform=ax2.transAxes)

# boxplot

ax3.boxplot(
    df["elevation_m"],
    vert=False,
    patch_artist=True,
    boxprops=dict(facecolor="red"),
    medianprops=dict(color="black"),
    whiskerprops=dict(color="black"),
    capprops=dict(color="black"),
    flierprops=dict(marker='o', markersize=6, markerfacecolor="white")
)

ax3.set_title("Diagrama de caja de la Altitud de las ciudades de California", fontsize=12)
ax3.set_xlabel("Altitud (m)")
ax3.set_yticks([1])
ax3.set_yticklabels([" "])  # Quitar etiqueta Y

# dispersion scatter

ax4.scatter(
    df["area_water_km2"],
    df["population_total"]/1000,
    color="green",
    alpha=0.6
)

ax4.set_title("Población por kilómetro cuadrado de agua", fontsize=12)
ax4.set_xlabel("Área de agua en la ciudad ($km^2$)")
ax4.set_ylabel("Población (miles habitantes)")

outliers = df[
    (df["area_water_km2"] > 80) &
    (df["population_total"] > 3000000)
]

# outliers
for _, row in outliers.iterrows():
    ax4.annotate(
        "Outliers",
        xy=(row["area_water_km2"], row["population_total"] / 1000),
        xytext=(row["area_water_km2"] - 30, row["population_total"] / 1000 + 500),
        arrowprops=dict(arrowstyle="->", lw=1),
        fontsize=10
    )

plt.tight_layout()

plt.show()




```
